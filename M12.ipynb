```python
class Karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.energi = 100
        self.gaji = gaji
    def tampilkan_gaji(self):
        print(f"Karyawan {self.nama} memiliki gaji {self.gaji}")
    def lembur(self):
        print(f"{self.nama} sedang lembur!")
        self.energi -= 30
        self.gaji += 200000

karyawan1 = Karyawan("Siti", 500000)
karyawan1.lembur()

print(f"Sisa energi {karyawan1.nama}: {karyawan1.energi}")
print(f"Gaji terbaru {karyawan1.nama}: {karyawan1.gaji}")
```

```python
class RekeningBank:
    def __init__(self, nama_pemilik, saldo=0):
        self.nama_pemilik = nama_pemilik
        self.saldo = saldo
    def menabung(self, nominal):
        self.saldo += nominal
        print(f"Berhasil menabung Rp{nominal}. Saldo sekarang: Rp{self.saldo}")
    def tarik_tunai(self, nominal):
        if self.saldo >= nominal:
            self.saldo -= nominal
            print(f"Berhasil menarik Rp{nominal}. Saldo sekarang: Rp{self.saldo}")
        else:
            print("Maaf, saldo tidak mencukupi.")

rekening_pengguna = RekeningBank("Anin", 150000)
rekening_pengguna.tarik_tunai(50000)
rekening_pengguna.tarik_tunai(200000)
rekening_pengguna.menabung(300000)
```
